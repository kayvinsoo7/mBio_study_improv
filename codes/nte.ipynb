## Explanation of the Revised Data Model

### 1. Fixed Average Energy by Task Type  
For each bioinformatics job type (e.g., GWAS, PRS, Alignment, etc.), we now fix the "Avg Energy Consumed (kWh)" value. For example, every GWAS task has an average energy consumption of 350 kWh irrespective of which consortium or researcher is involved. This allows us to compare rows of the same task type meaningfully.

**Example Values by Task:**

- PRS: 280 kWh  
- Alignment: 420 kWh  
- Variant Calling: 310 kWh

- Metagenomics: 500 kWh  
- RNA‑seq: 300 kWh  
- Proteomics: 450 kWh  
- Epigenomics: 330 kWh  
- Functional Annotation: 290 kWh  

### 2. The Role of Tasks Rerun and Subtasks Delegated  
- **Tasks Rerun:**  
  If a task has to be re-run, it means the work is repeated (perhaps because of errors or to improve quality) and this adds extra energy demand. We model this by increasing the energy consumption in the “Baseline Energy” calculation. For instance, we apply a factor such as a 2% increase per task re-run:
  
  \[
  \text{Baseline Energy} = \text{Avg Energy} \times \left(1 + 0.02 \times \text{(Tasks Rerun)}\right)
  \]
  
- **Subtasks Delegated:**  
  In our conceptual framework, delegating subtasks to a data manager implies that parts of the computational pipeline are offloaded, thus reducing the energy consumed. A higher number in this column means more of the pipeline is handled by a more efficient process. We model that as a percentage reduction in energy for each delegated subtask. For example, using a 3% reduction per subtask, we get:
  
  \[
  \text{Optimized Energy} = \text{Baseline Energy} \times \left(1 - 0.03 \times \text{(Subtasks Delegated)}\right)
  \]
  
  (A safeguard should be in place to avoid negative energy, but with our chosen numbers, the reduction stays within reasonable limits.)

### 3. Calculation of Other Columns  
Based on the above two calculations, we derive the remaining columns:
  
- **Runtime Increase (%):**  
  Delegating subtasks introduces an overhead. We assume a flat overhead percentage per subtask delegated. Here, we use 5% per subtask; that is:
  
  \[
  \text{Runtime Increase (\%)} = 5 \times \text{(Subtasks Delegated)}
  \]
  
- **CO₂ Saved (kg):**  
  The difference between the baseline and optimized energy consumption is “saved” energy. We convert this to estimated CO₂ savings by assuming a conversion factor—in this case, 0.5 kg CO₂ saved per kWh saved:
  
  \[
  \text{CO₂ Saved (kg)} = \left(\text{Baseline Energy} - \text{Optimized Energy}\right) \times 0.5
  \]
  
- **Trees to Offset Energy:**  
  We assume one tree can offset roughly 50 kWh. Thus, the number of trees required to offset consumption is given by:
  
  \[
  \text{Trees to Offset} = \text{Round}\left(\frac{\text{Energy Value}}{50}\right)
  \]
  
  The table includes two such columns (one for the baseline and one for the optimized energy values).

### **Abstract**

**Background:**  
Early identification of individuals at high risk for Type 2 Diabetes (T2D) is critical for timely intervention. However, many cases remain undiagnosed despite clinical eligibility. This study aimed to identify high-risk subgroups within a population and assess the proportion of undiagnosed T2D cases within each.

**Methods:**  
We applied a scanning algorithm across various penalty parameters to identify compact and high-risk subpopulations based on behavioral, clinical, and demographic features. For each identified subgroup, we calculated the odds ratio for T2D, subset size, diabetes case concentration, and the proportion of individuals meeting clinical T2D thresholds who did not self-report the condition (i.e., undiagnosed cases).

**Results:**  
We found several subgroups with substantially elevated odds of T2D (up to 10.7×), concentrated diabetes cases (up to 46.95%), and varying levels of undiagnosed T2D. Despite the high-risk nature of these groups, undiagnosed T2D rates remained relatively low (0.97–1.6%), suggesting effective case capture in core risk subpopulations. However, certain subgroups with simplified definitions and moderate odds still exhibited elevated undiagnosed rates, indicating potential targets for screening outreach.

**Conclusion:**  
This subgroup-based approach reveals that while high-risk groups are largely diagnosed, opportunities remain to improve detection among phenotypically identifiable clusters. These findings can inform more targeted and equitable diabetes screening strategies.
